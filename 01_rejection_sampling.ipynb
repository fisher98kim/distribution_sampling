```python
# --------------------
# Pseudo
# --------------------

'''
1. find M, that target_dist(x) <= M*proposal_dist(x) for all x
   - M = sup_x target_dist(x) / i_proposal_dist(x)

2. repeat until N samples accepted:
   a. x ~ i_proposal_sample()
   b. u ~ Uniform(0, 1)
   c. if u <= target_dist(x) / (M * i_proposal_dist(x)):
          accept x
      else:
          reject x
'''
```

```python
# --------------------
# Example
# --------------------
min_x, max_x = -5, 5


def target_dist(x):
    return stats.norm.pdf(x, loc=0, scale=1)


def i_proposal_dist(x):
    # --------------------
    # Independent Proposal Distribution
    # --------------------
    # stats.uniform.pdf(x, -5, 10) = Uniform(-5,5)
    return stats.uniform.pdf(x, min_x, max_x-min_x)


def i_proposal_sample():
    # --------------------
    # Independent Proposal Sampler
    # --------------------
    return np.random.uniform(min_x, max_x)
```

```python
# find M
Xs = np.linspace(min_x, max_x, 10_000)

target_vals = target_dist(Xs)
proposal_vals = i_proposal_dist(Xs)

M = max(target_vals/proposal_vals)
```

```python
samples = []
n_accept = 0
n_reject = 0


while len(samples) < 10_000:

    x = i_proposal_sample()
    u = np.random.uniform(0, 1)
    if u <= target_dist(x) / (M * i_proposal_dist(x)):
        samples.append(x)
        n_accept += 1
    else:
        n_reject += 1
```

```python
plt.plot(Xs, target_vals)
plt.hist(samples, density=True, bins=100)
plt.tight_layout()
plt.show()
```
