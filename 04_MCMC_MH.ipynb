```python
# --------------------
# Example
# --------------------
sigma = 1


def target_dist(x):
    return stats.norm.pdf(x, loc=0, scale=1)


def d_proposal_dist(x, x0):
    # --------------------
    # Dependent Proposal Distribution
    # --------------------
    return stats.uniform.pdf(x, x0-sigma, 2*sigma)


def d_proposal_sample(x0):
    # --------------------
    # Dependent Proposal Sampler
    # --------------------
    return np.random.uniform(x0-sigma, x0+sigma)
```

```python
x_init = 0
x_curr = x_init
samples = [x_curr]
while len(samples) < 10_000:

    x_proposal = d_proposal_sample(x_curr)

    p_curr = target_dist(x_curr)
    p_proposal = target_dist(x_proposal)

    q_forward = d_proposal_dist(x_proposal, x_curr)
    q_backward = d_proposal_dist(x_curr, x_proposal)

    alpha = min(1.0, (p_proposal*q_backward) / (p_curr * q_forward))

    u = np.random.uniform(0, 1)
    if u <= alpha:
        x_curr = x_proposal
    else:
        x_curr = x_curr
    
    samples.append(x_curr)
```

```python
Xs = np.linspace(-5, 5, 10_000)
target_vals = target_dist(Xs)

plt.plot(Xs, target_vals)
plt.hist(samples, density=True, bins=100)
plt.tight_layout()
plt.show()
```
