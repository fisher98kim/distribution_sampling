```python
# --------------------
# Example
# --------------------
sigma = 5


def target_dist(x):
    return stats.norm.pdf(x, loc=0, scale=1)


def i_proposal_dist(x):
    # --------------------
    # Independent Proposal Distribution
    # --------------------
    return stats.uniform.pdf(x, 0-sigma, 2*sigma)


def i_proposal_sample():
    # --------------------
    # Independent Proposal Sampler
    # --------------------
    return np.random.uniform(0-sigma, 0+sigma)
```

```python
x_init = 0
x_current = x_init
samples = [x_current]
while len(samples) < 10_000:

    x_proposal = i_proposal_sample()

    p_current = target_dist(x_current)
    p_proposal = target_dist(x_proposal)

    q_current = i_proposal_dist(x_current)
    q_proposal = i_proposal_dist(x_proposal)

    w_current = p_current / q_current
    w_proposal = p_proposal / q_proposal
    alpha = min(1.0, w_proposal / w_current)

    u = np.random.uniform(0, 1)
    if u <= alpha:
        x_current = x_proposal
    else:
        x_current = x_current

    samples.append(x_current)
```

```python
Xs = np.linspace(-5, 5, 10_000)
target_vals = target_dist(Xs)

plt.plot(Xs, target_vals)
plt.hist(samples, density=True, bins=100)
plt.tight_layout()
plt.show()
```
